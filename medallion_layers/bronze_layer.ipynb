## Transfering Ingested Data by ADF to Bronze Layer in their raw format

```python
# Load raw data as Parquet files ingested from sql
df_sql= spark.read.format("parquet") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/mnt/raw/*.parquet")
# Load raw data as csv files ingested from url
df_csv= spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/mnt/raw/*.csv")

display(df_csv)
display(df_sql)    

```

## Mounting Bronze Layer Container for storing raw ingested Data 

```python

container = "bronze-layer"
storage_account = "etlprojectspotify"
mount_name = "bronze"

result=dbutils.notebook.run("/Workspace/Users/suman.kr.ghorai@gmail.com/utils/mount_utils", 60, {
    "container": container,
    "storage_account": storage_account,
    "mount_name": mount_name
})
print(result)
```

```python
dbutils.fs.ls("/mnt/bronze")
```

## Saving 
-      Data In Bronze Layer
    

```python
# # Save csv data to bronze layer (in original format)
df_csv.write.format("csv") \
    .option("header", "true") \
    .mode("overwrite") \
    .save("/mnt/bronze/csv_data")

# Save Parquet data to bronze layer (in original format)
df_sql.write.format("parquet") \
    .mode("overwrite") \
    .save("/mnt/bronze/sql_data")
```
